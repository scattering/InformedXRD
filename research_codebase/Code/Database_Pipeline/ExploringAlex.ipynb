```python
json_path = 'JSONS\\PBE\\3D_001.json'
```

```python
#Analyze Keys
def analyze_json_structure(json_data, indent=0, analyzed=None):
    """Analyzes the structure of the given JSON data."""
    
    if analyzed is None:
        analyzed = set()
    
    # Check if the JSON data is a dictionary
    if isinstance(json_data, dict):
        obj_id = id(json_data)
        if obj_id in analyzed:
            print("  " * indent + "Dictionary (already analyzed)")
            return
        analyzed.add(obj_id)
        
        print("  " * indent + "Dictionary with the following keys:")
        for key in json_data.keys():
            print("  " * (indent + 1) + f"- {key}")
        input("Press Enter to continue...")
        for key, value in json_data.items():
            if isinstance(value, (dict, list)):
                print("  " * (indent + 1) + f"Analyzing '{key}'...")
                analyze_json_structure(value, indent + 2, analyzed)
    
    # Check if the JSON data is a list
    elif isinstance(json_data, list):
        obj_id = id(json_data)
        if obj_id in analyzed:
            print("  " * indent + "List (already analyzed)")
            return
        analyzed.add(obj_id)
        
        print("  " * indent + "List")
        input("Press Enter to continue...")
        if len(json_data) > 0:
            first_element = json_data[0]
            if isinstance(first_element, (dict, list)):
                print("  " * (indent + 1) + "First element is a:")
                analyze_json_structure(first_element, indent + 2, analyzed)

try:
    with open(json_path, 'r') as file:
        json_data = json.load(file)
        analyze_json_structure(json_data)
except FileNotFoundError:
    print("File not found. Please check the file path.")
except json.JSONDecodeError as e:
    print(f"Failed to parse JSON: {e}")
```

```python
import json
import pandas

paths = ['alexandria_ps_000.json\\alexandria_ps_000.json']
formulas = []
spacegroups = []
for path in paths:
    x = 0
    with open(path, 'r') as file:
        json_data = json.load(file)
        for entry in json_data['entries']:
            x+=1
            formulas.append(entry['data']['formula'])
            spacegroups.append(entry['data']['spg'])
            if(x > 10):
                break


df = pandas.DataFrame(data={"Formula":formulas,"Space Group #":spacegroups})
df.to_csv("Test CSV.csv",index=False)
```
