```python
import numpy as np

def load_and_format_data(data_source):
    """
    Load and format numeric data into a NumPy array, replacing NaN values with 0.

    Parameters
    ----------
    data_source : str or list or np.ndarray
        - If str: expects a file path or a string containing the data.
        - If list or np.ndarray: directly processes the data.

    Returns
    -------
    np.ndarray
        A 1D NumPy array with NaN values replaced by 0.
    """
    # Case 1: If input is a string pointing to a file
    if isinstance(data_source, str):
        try:
            # Try loading from file first
            data = np.loadtxt(data_source)
        except OSError:
            # Otherwise, treat as raw text data (string of numbers)
            data_str = data_source.strip().replace('\n', ' ')
            data = np.fromstring(data_str, sep=' ')
    # Case 2: If input is already a list or ndarray
    elif isinstance(data_source, (list, np.ndarray)):
        data = np.array(data_source, dtype=float)
    else:
        raise TypeError("Unsupported data_source type. Must be str, list, or np.ndarray.")

    # Replace NaN with 0
    data = np.nan_to_num(data, nan=0.0)

    return data
```
